## **Class-Conditional Image Generation with Conditional Generative Adversarial Networks (cGANs)**

````python
# run on GPU (Colab)
# !pip install torchmetrics[image]
import os
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchvision
from torchvision import transforms
from torchvision.utils import make_grid
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
import imageio
from tqdm import tqdm
from torchmetrics.image.fid import FrechetInceptionDistance

LATENT_DIM = 128
NUM_CLASSES = 10
BATCH_SIZE = 128
EPOCHS = 80

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
os.makedirs("dashboard", exist_ok=True)
os.makedirs("models", exist_ok=True)
os.makedirs("videos", exist_ok=True)
os.makedirs("samples", exist_ok=True)
transform = transforms.Compose([transforms.ToTensor(), transforms.Normalize((0.5,), (0.5,))])
dataset = torchvision.datasets.FashionMNIST("./data", train=True, download=True, transform=transform)
loader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=2)

class Generator(nn.Module):
    def __init__(self):
        super().__init__()
        self.emb = nn.Embedding(NUM_CLASSES, 128)

        self.fc = nn.Sequential(
            nn.Linear(LATENT_DIM + 128, 7*7*256),
            nn.BatchNorm1d(7*7*256),
            nn.ReLU()
        )

        self.net = nn.Sequential(
            nn.ConvTranspose2d(256,128,4,2,1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.ConvTranspose2d(128,64,4,2,1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.Conv2d(64,1,3,1,1),
            nn.Tanh()
        )

    def forward(self,z,y):
        y = self.emb(y)
        x = torch.cat([z,y],dim=1)
        x = self.fc(x).view(-1,256,7,7)
        return self.net(x)

class Discriminator(nn.Module):
  def __init__(self):
      super().__init__()
      self.emb = nn.Embedding(NUM_CLASSES, 256)

      self.conv = nn.Sequential(
          nn.utils.spectral_norm(nn.Conv2d(1,64,4,2,1)),
          nn.LeakyReLU(0.2),
          nn.utils.spectral_norm(nn.Conv2d(64,128,4,2,1)),
          nn.LeakyReLU(0.2),
          nn.utils.spectral_norm(nn.Conv2d(128,256,3,1,1)),
          nn.LeakyReLU(0.2),
      )

      self.fc = nn.utils.spectral_norm(nn.Linear(256*7*7,256))
      self.out = nn.utils.spectral_norm(nn.Linear(256,1))

  def forward(self,x,y):
      h = self.conv(x).view(x.size(0),-1)
      h = self.fc(h)
      out = self.out(h)
      proj = (h * self.emb(y)).sum(1, keepdim=True)
      return out + proj

G = Generator().to(device)
D = Discriminator().to(device)
G_ema = Generator().to(device)
G_ema.load_state_dict(G.state_dict())
for p in G_ema.parameters():
    p.requires_grad = False

def ema_update(beta=0.999):
    with torch.no_grad():
        for p, p_ema in zip(G.parameters(), G_ema.parameters()):
            p_ema.data.mul_(beta).add_(p.data*(1-beta))

opt_g = torch.optim.Adam(G.parameters(), lr=2e-4, betas=(0.0,0.9))
opt_d = torch.optim.Adam(D.parameters(), lr=4e-4, betas=(0.0,0.9))
sched_g = torch.optim.lr_scheduler.CosineAnnealingLR(opt_g, EPOCHS)
sched_d = torch.optim.lr_scheduler.CosineAnnealingLR(opt_d, EPOCHS)
fid_metric = FrechetInceptionDistance(feature=64).to(device)

def to_fid_format(x):
    """
    Convert [-1,1] float tensor → uint8 [0,255]
    and ensure 3 channels
    """
    x = (x + 1) / 2          # [-1,1] → [0,1]
    x = x.clamp(0, 1)
    x = x * 255.0
    x = x.to(torch.uint8)
    if x.shape[1] == 1:
        x = x.repeat(1, 3, 1, 1)
    return x

def compute_fid(real, fake):
    fid_metric.reset()
    fid_metric.update(real, real=True)
    fid_metric.update(fake, real=False)
    return fid_metric.compute().item()

g_losses, d_losses, fid_scores = [], [], []

def update_dashboard(epoch):
    plt.figure(figsize=(12,4))

    plt.subplot(1,3,1)
    plt.plot(g_losses, label="G")
    plt.plot(d_losses, label="D")
    plt.legend()
    plt.title("Losses")
    plt.subplot(1,3,2)
    plt.plot(fid_scores, label="FID")
    plt.title("FID Curve")
    plt.subplot(1,3,3)
    z = torch.randn(100, LATENT_DIM, device=device)
    y = torch.arange(10).repeat_interleave(10).to(device)

    with torch.no_grad():
        img = G_ema(z,y)
    grid = make_grid(img, nrow=10, normalize=True)
    plt.imshow(np.transpose(grid.cpu(),(1,2,0)), cmap="gray")
    plt.title("Samples")
    plt.tight_layout()
    #plt.savefig(f"dashboard/epoch_{epoch}.png")
    plt.close()

best_fid = 1e9

def save_best(fid_score):
    global best_fid
    if fid_score < best_fid:
        best_fid = fid_score
        torch.save(G_ema.state_dict(), "models/G_ema.pth")
        print("🔥 NEW BEST MODEL SAVED")

fixed_z = torch.randn(100, LATENT_DIM, device=device)
fixed_y = torch.arange(10).repeat_interleave(10).to(device)

for epoch in range(EPOCHS):
    loop = tqdm(loader)
    for real,y in loop:
        real,y = real.to(device), y.to(device)
        # D
        z = torch.randn(real.size(0), LATENT_DIM, device=device)
        fake = G(z,y).detach()
        d_loss = (F.relu(1-D(real,y)).mean() + F.relu(1+D(fake,y)).mean())
        opt_d.zero_grad()
        d_loss.backward()
        opt_d.step()
        # G
        z = torch.randn(real.size(0), LATENT_DIM, device=device)
        fake = G(z,y)
        g_loss = -D(fake,y).mean()
        opt_g.zero_grad()
        g_loss.backward()
        opt_g.step()
        ema_update()
    sched_g.step()
    sched_d.step()

    # ---------------- FID ----------------
    with torch.no_grad():
        z = torch.randn(200, LATENT_DIM, device=device)
        y = torch.randint(0,10,(200,),device=device)
        fake = G_ema(z,y)
        real_batch = next(iter(loader))[0][:200].to(device)
        real_fid = to_fid_format(real_batch)
        fake_fid = to_fid_format(fake)
        fid_score = compute_fid(real_fid, fake_fid)

    fid_scores.append(fid_score)
    # save losses
    g_losses.append(g_loss.item())
    d_losses.append(d_loss.item())
    update_dashboard(epoch)

    # save samples
    with torch.no_grad():
        img = G_ema(fixed_z,fixed_y)
        grid = make_grid(img, nrow=10, normalize=True)
        plt.imshow(np.transpose(grid.cpu(),(1,2,0)), cmap="gray")
        plt.axis("off")
        #plt.savefig(f"samples/{epoch}.png")
        plt.close()
    save_best(fid_score)
    print(f"E{epoch} | G:{g_loss.item():.3f} | D:{d_loss.item():.3f} | FID:{fid_score:.2f}")
````

````python
classes = ["T-shirt", "Trouser", "Pullover", "Dress", "Coat", "Sandal", "Shirt", "Sneaker", "Bag", "Ankle Boot"]
G.eval()

def class_transition(class_a, class_b, steps=15):
    z = torch.randn(1, LATENT_DIM, device=device)
    emb_a = G.emb(torch.tensor([class_a], device=device))
    emb_b = G.emb(torch.tensor([class_b], device=device))
    imgs = []
    with torch.no_grad():
        for alpha in np.linspace(0,1,steps):
            emb = ((1-alpha)*emb_a + alpha*emb_b)
            img = generate_from_embedding(G, z, emb)
            imgs.append(img.cpu())
    return imgs

def generate_from_embedding(G, z, emb):
    x = torch.cat([z, emb], dim=1)
    x = G.fc(x)
    x = x.view(-1, 256, 7, 7)
    return G.net(x)

images = class_transition(7, 9, steps=15)
fig, axes = plt.subplots(1, len(images), figsize=(20,3))
for ax,img in zip(axes,images):
    img = (img[0,0].numpy())
    ax.imshow(img, cmap="gray")
    ax.axis("off")
plt.suptitle("Sneaker → Ankle Boot")
plt.show()
````

````python
G.eval()
# Choose four Fashion-MNIST classes
top_left = 0      # T-shirt
top_right = 4     # Coat
bottom_left = 7   # Sneaker
bottom_right = 9  # Ankle Boot
# Fixed latent vector
z = torch.randn(1, LATENT_DIM, device=device)
emb_tl = G.emb(torch.tensor([top_left], device=device))
emb_tr = G.emb(torch.tensor([top_right], device=device))
emb_bl = G.emb(torch.tensor([bottom_left], device=device))
emb_br = G.emb(torch.tensor([bottom_right], device=device))
grid_size = 10
fig, axes = plt.subplots(grid_size, grid_size, figsize=(6,6))

with torch.no_grad():
    for i in range(grid_size):
        y = i/(grid_size-1)
        for j in range(grid_size):
            x = j/(grid_size-1)
            # Bilinear interpolation of embeddings
            emb = ((1-x)*(1-y)*emb_tl + x*(1-y)*emb_tr + (1-x)*y*emb_bl + x*y*emb_br)
            inp = torch.cat([z, emb], dim=1)
            feat = G.fc(inp)
            feat = feat.view(-1, 256, 7, 7)
            img = G.net(feat)
            img = ((img[0,0]+1)/2).cpu().numpy()
            axes[i,j].imshow(img, cmap="gray")
            axes[i,j].axis("off")

plt.subplots_adjust(wspace=0.02, hspace=0.02)
plt.suptitle("2D Fashion-MNIST Class Manifold", fontsize=16)
plt.show()
````

## **Conditional Image Generation with Conditional Variational AutoEncoder (CVAE)**

````python
# run in colab / GPU
import os
import imageio
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchvision
from torchvision import transforms
from torchvision.utils import make_grid
from torch.utils.data import DataLoader
from tqdm import tqdm
from sklearn.manifold import TSNE

LATENT_DIM = 64
NUM_CLASSES = 10
BATCH_SIZE = 128
EPOCHS = 80
LR = 1e-3
BETA = 0.5

device = torch.device( "cuda" if torch.cuda.is_available() else "cpu")
print(device)
os.makedirs("checkpoints", exist_ok=True)
os.makedirs("samples", exist_ok=True)
os.makedirs("manifold", exist_ok=True)
os.makedirs("videos", exist_ok=True)

transform = transforms.Compose([transforms.ToTensor()])
dataset = torchvision.datasets.FashionMNIST("./data", train=True, download=True, transform=transform)
loader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=2, pin_memory=True)
class_names = ["T-shirt", "Trouser", "Pullover", "Dress", "Coat", "Sandal", "Shirt", "Sneaker", "Bag", "Boot"]

class CVAE(nn.Module):

    def __init__(self):

        super().__init__()
        
        # CLASS CONDITIONING 
        self.label_fc = nn.Linear(NUM_CLASSES, 32)
        
        # ENCODER
        self.encoder = nn.Sequential(
            nn.Linear(784 + 32, 512),
            nn.ReLU(),
            nn.Linear(512, 256),
            nn.ReLU()
        )

        self.mu = nn.Linear(256, LATENT_DIM)
        self.logvar = nn.Linear(256, LATENT_DIM)

        # DECODER
        self.decoder = nn.Sequential(
            nn.Linear(LATENT_DIM + 32, 256),
            nn.ReLU(),
            nn.Linear(256, 512),
            nn.ReLU(),
            nn.Linear(512, 784),
            nn.Sigmoid()
        )

    def encode(self, x, y):
        """
        x: (B, 784)
        y: (B,) integer class labels
        """
        # convert to one-hot
        y = F.one_hot(y, NUM_CLASSES).float()
        # continuous conditioning
        y = self.label_fc(y)
        h = torch.cat([x, y], dim=1)
        h = self.encoder(h)
        return self.mu(h), self.logvar(h)

    def reparameterize(self, mu, logvar):
        logvar = torch.clamp(logvar, -10, 10)
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std

    def decode(self, z, y):
        """
        z: (B, LATENT_DIM)
        y: (B,) integer labels OR (B, NUM_CLASSES) soft labels
        """
        # allow both formats (IMPORTANT for manifold work)
        if y.dim() == 1 or y.shape[-1] != NUM_CLASSES:
            y = F.one_hot(y.long(), NUM_CLASSES).float()
        y = self.label_fc(y)
        h = torch.cat([z, y], dim=1)
        return self.decoder(h)

    def forward(self, x, y):
        mu, logvar = self.encode(x, y)
        z = self.reparameterize(mu, logvar)
        recon = self.decode(z, y)
        return recon, mu, logvar

model = CVAE().to(device)

def cvae_loss(recon, x, mu, logvar):

    recon = torch.clamp(recon, min=1e-7, max=1-1e-7)
    recon_loss = F.binary_cross_entropy(recon, x, reduction="sum")
    kl = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())
    total = (recon_loss + BETA * kl)
    return (total, recon_loss, kl)

optimizer = torch.optim.Adam(model.parameters(), lr=LR)
CHECKPOINT_FILE = "models/cvae_latest.pth"
start_epoch = 0
loss_history = []
recon_history = []
kl_history = []
best_loss = float("inf")

if os.path.exists(CHECKPOINT_FILE):
    ckpt = torch.load(CHECKPOINT_FILE, map_location=device)
    model.load_state_dict(ckpt["model"])
    optimizer.load_state_dict(ckpt["optimizer"])
    start_epoch = ckpt["epoch"]
    loss_history = ckpt["loss_history"]
    recon_history = ckpt["recon_history"]
    kl_history = ckpt["kl_history"]
    best_loss = ckpt["best_loss"]
    print(f"Resuming from epoch {start_epoch}")

fixed_z = torch.randn(100, LATENT_DIM, device=device)
fixed_labels = torch.arange(10, device=device).repeat_interleave(10)

def save_epoch_samples(epoch):
    
    model.eval()

    with torch.no_grad():
        fake = model.decode(fixed_z, fixed_labels)
        fake = fake.view(-1, 1, 28, 28)
        grid = make_grid(fake, nrow=10, normalize=True)
        plt.figure(figsize=(8,8))
        plt.imshow(grid.cpu().permute(1,2,0), cmap="gray")
        plt.axis("off")
        plt.tight_layout()
        #plt.savefig(f"samples/epoch_{epoch:03d}.png")
        plt.close()

for epoch in range(start_epoch, EPOCHS):

    model.train()
    total_loss = 0
    total_recon = 0
    total_kl = 0
    loop = tqdm(loader, leave=False)

    for images, labels in loop:
        images = images.to(device)
        labels = labels.to(device)
        x = images.view(images.size(0), -1)
        recon, mu, logvar = model(x, labels)
        loss, recon_loss, kl = cvae_loss(recon, x, mu, logvar)
        optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)
        optimizer.step()
        total_loss += loss.item()
        total_recon += recon_loss.item()
        total_kl += kl.item()

    avg_loss = total_loss / len(dataset)
    avg_recon = total_recon / len(dataset)
    avg_kl = total_kl / len(dataset)
    loss_history.append(avg_loss)
    recon_history.append(avg_recon)
    kl_history.append(avg_kl)
    print(f"Epoch {epoch+1:03d} | "
        f"Loss={avg_loss:.4f} | "
        f"Recon={avg_recon:.4f} | "
        f"KL={avg_kl:.4f}")
    save_epoch_samples(epoch+1)

    if avg_loss < best_loss:
        best_loss = avg_loss
        torch.save(model.state_dict(), "models/cvae_best.pth")

    torch.save({
            "epoch": epoch+1,
            "model": model.state_dict(),
            "optimizer": optimizer.state_dict(),
            "loss_history": loss_history,
            "recon_history": recon_history,
            "kl_history": kl_history,
            "best_loss": best_loss
        }, CHECKPOINT_FILE)

plt.figure(figsize=(10,5))
plt.plot(loss_history, label="Total")
plt.plot(recon_history, label="Recon")
plt.plot(kl_history, label="KL")
plt.legend()
plt.title("CVAE Training Curves")
#plt.savefig("images/loss_curve.png")
plt.show()

model.eval()

fig, axes = plt.subplots(2, 5, figsize=(10,5))
with torch.no_grad():
    for cls in range(10):
        z = torch.randn(1, LATENT_DIM, device=device)
        label = torch.tensor([cls], device=device)
        img = model.decode(z, label)
        img = img.view(28, 28).cpu()
        r = cls // 5
        c = cls % 5
        axes[r,c].imshow(img, cmap="gray")
        axes[r,c].set_title(class_names[cls])
        axes[r,c].axis("off")
plt.tight_layout()
#plt.savefig("images/cvae_class_samples.png")
plt.show()

#frames = []
#files = sorted(glob.glob("samples/*.png"))
#for f in files:
#    frames.append(imageio.imread(f))
#imageio.mimsave("images/cvae_train.gif", frames, fps=5)
````

````python
model.eval()
features = []
labels_all = []
counts = {i:0 for i in range(10)}

with torch.no_grad():
    
    for images, labels in loader:
        images = images.to(device)
        x = images.view(images.size(0), -1)
        mu, _ = model.encode(x, labels.to(device))
        for z,lbl in zip(mu.cpu(), labels):
            cls = lbl.item()
            if counts[cls] < 10:
                features.append(z.numpy())
                labels_all.append(cls)
                counts[cls] += 1
        if min(counts.values()) >= 10:
            break

features = np.array(features)
coords = TSNE(n_components=2, perplexity=20, random_state=42).fit_transform(features)

plt.figure(figsize=(8,8))
scatter = plt.scatter(coords[:,0], coords[:,1], c=labels_all, cmap="tab10", s=80)
plt.colorbar(scatter)
plt.title("CVAE Latent Space (t-SNE)")
#plt.savefig("images/tsne_latent.png")
plt.show()

from matplotlib.offsetbox import (OffsetImage, AnnotationBbox)

thumbs = []
counts = {i:0 for i in range(10)}
with torch.no_grad():
    for images, labels in loader:
        for img,lbl in zip(images, labels):
            cls = lbl.item()
            if counts[cls] < 10:
                thumbs.append(img.squeeze().numpy())
                counts[cls]+=1
        if min(counts.values())>=10:
            break

fig, ax = plt.subplots(figsize=(12,12))
ax.scatter(coords[:,0], coords[:,1], alpha=0)
for xy, img in zip(coords, thumbs):
    ab = AnnotationBbox(OffsetImage(img, zoom=0.7, cmap="gray"), xy, frameon=False)
    ax.add_artist(ab)
plt.title("CVAE Latent Space with Fashion-MNIST Samples")
#plt.savefig("images/tsne_images.png")
plt.show()

# latent (within class) traveral: changes styles
img_sneaker = None
img_bag = None
for img,lbl in dataset:
    if lbl == 7 and img_sneaker is None:
        img_sneaker = img
    if lbl == 8 and img_bag is None:
        img_bag = img
    if (img_sneaker is not None andimg_bag is not None):
        break

with torch.no_grad():
    x1 = img_sneaker.view(1,-1).to(device)
    x2 = img_bag.view(1,-1).to(device)
    y1 = torch.tensor([7], device=device)
    y2 = torch.tensor([8], device=device)
    mu1,_ = model.encode(x1, y1)
    mu2,_ = model.encode(x2, y2)

fig, axes = plt.subplots(1, 15,figsize=(18,3))
with torch.no_grad():
    for i,a in enumerate(np.linspace(0,1,15)):
        z = ((1-a)*mu1 + a*mu2)
        out = model.decode(z, y1)
        img = out.view(28, 28).cpu()
        axes[i].imshow(img, cmap="gray")
        axes[i].axis("off")
#plt.savefig("images/latent_traversal.png")
plt.show()

class_latents = {i:[] for i in range(10)}

with torch.no_grad():
    for images, labels in loader:
        images = images.to(device)
        x = images.view(images.size(0), -1)
        mu,_ = model.encode(x, labels.to(device))
        for z,lbl in zip(mu.cpu(), labels):
            class_latents[lbl.item()].append(z)

# 1. Collect latent means per class
class_latents = {i: [] for i in range(10)}
with torch.no_grad():
    for images, labels in loader:
        images = images.to(device)
        x = images.view(images.size(0), -1)
        mu, _ = model.encode(x, labels.to(device))
        for z, lbl in zip(mu.cpu(), labels):
            class_latents[lbl.item()].append(z)

class_means = {
    cls: torch.stack(class_latents[cls]).mean(0)
    for cls in range(10)
}

# 2. Choose 4 corner classes
cls_tl = 8
cls_tr = 7
cls_bl = 3
cls_br = 9
z_tl = class_means[cls_tl]
z_tr = class_means[cls_tr]
z_bl = class_means[cls_bl]
z_br = class_means[cls_br]

# one-hot labels
def one_hot(cls, num_classes=10):
    v = torch.zeros(num_classes)
    v[cls] = 1.0
    return v

y_tl = one_hot(cls_tl)
y_tr = one_hot(cls_tr)
y_bl = one_hot(cls_bl)
y_br = one_hot(cls_br)

# 3. Grid setup
GRID = 12
H, W = 28, 28
canvas = np.zeros((GRID * H, GRID * W))
# 4. Manifold traversal
model.eval()
with torch.no_grad():
    for i in range(GRID):          # vertical (v)
        v = i / (GRID - 1)
        for j in range(GRID):      # horizontal (u)
            u = j / (GRID - 1)
            # bilinear weights
            w_tl = (1 - u) * (1 - v)
            w_tr = u * (1 - v)
            w_bl = (1 - u) * v
            w_br = u * v
            # interpolate latent
            z = (w_tl * z_tl + w_tr * z_tr + w_bl * z_bl + w_br * z_br).unsqueeze(0).to(device)
            # interpolate label (CRITICAL FIX)
            y = (w_tl * y_tl + w_tr * y_tr + w_bl * y_bl + w_br * y_br).unsqueeze(0).to(device)
            # decode
            img = model.decode(z, y)
            img = img.view(H, W).cpu().numpy()
            canvas[i*H:(i+1)*H, j*W:(j+1)*W] = img

# 5. Visualization
plt.figure(figsize=(10, 10))
plt.imshow(canvas, cmap="gray")
plt.axis("off")
plt.title("CVAE 4-Corner Conditional Latent Manifold (z + label interpolation)")
#plt.savefig("images/cvae_latent2D.png") # four_corner_manifold
plt.show()

z = torch.randn(1, LATENT_DIM, device=device)
fig, axes = plt.subplots(2, 5, figsize=(12,5))
with torch.no_grad():
    for cls in range(10):
        label = torch.tensor([cls], device=device)
        img = model.decode(z, label)
        img = img.view(28, 28).cpu()
        r = cls // 5
        c = cls % 5
        axes[r,c].imshow(img, cmap="gray")
        axes[r,c].set_title(class_names[cls])
        axes[r,c].axis("off")
plt.tight_layout()
#plt.savefig("images/same_z_different_labels.png")
plt.show()
````

### **StyleGANEX: High-Fidelity Inversion and Identity Editing**

````python
# run in Colab GPU / TPU
# INSTALL / SETUP
!git clone https://github.com/williamyang1991/StyleGANEX.git
%cd StyleGANEX

import torch, cv2, imageio, numpy as np
from PIL import Image
from tqdm import tqdm
import dlib
from models.psp import pSp
from argparse import Namespace
from huggingface_hub import hf_hub_download
from scripts.align_all_parallel import align_face
from latent_optimization import latent_optimization
from utils.inference_utils import get_video_crop_parameter


# =========================
# LOAD MODEL (CLEAN)
# =========================
def load_model(path, device="cuda"):
    ckpt_path = hf_hub_download("PKUWilliamYang/StyleGANEX", path)
    ckpt = torch.load(ckpt_path, map_location="cpu")

    opts = Namespace(**ckpt["opts"])
    opts.checkpoint_path = ckpt_path
    opts.device = device

    model = pSp(opts).to(device).eval()
    model.latent_avg = model.latent_avg.to(device)

    return model


psp = load_model("pretrained_models/styleganex_inversion.pt", "cuda")


# =========================
# IMAGE PROCESSING
# =========================
def load_crop(img_path, lm):
    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    h, w, t, b, l, r, _ = get_video_crop_parameter(img, lm)
    return cv2.resize(img, (w, h))[t:b, l:r]


# =========================
# W+ INVERSION
# =========================
def invert(img, model, lm, steps=500):
    w_hat, f_hat, noise_hat, w, f = latent_optimization(
        img, model, lm, step=steps, device="cuda"
    )
    return w_hat, f_hat, noise_hat, w, f


# =========================
# INTERPOLATION + RENDER
# =========================
def decode(model, w, f, noise):
    with torch.no_grad():
        img, _ = model.decoder([w], input_is_latent=True,
                               first_layer_feature=f,
                               noise=noise)
    img = (img.clamp(-1,1) + 1) * 127.5
    img = img.permute(0,2,3,1)[0].cpu().numpy().astype(np.uint8)
    return Image.fromarray(img)


def morph(img1_path, img2_path, lm, steps=30, inv_steps=500):
    img1 = load_crop(img1_path, lm)
    img2 = load_crop(img2_path, lm)

    w1, f1, n1, w1c, f1c = invert(img1, psp, lm, inv_steps)
    w2, f2, n2, w2c, f2c = invert(img2, psp, lm, inv_steps)

    frames = []

    for a in tqdm(np.linspace(0,1,steps)):
        w = (1-a)*w1 + a*w2
        f = f1
        n = n1

        frames.append(np.array(decode(psp, w, f, n)))

    imageio.mimsave("morph.gif", frames, fps=15)
    print("Saved → morph.gif")


# =========================
# LANDMARK MODEL
# =========================
landmarkpredictor = dlib.shape_predictor(
    hf_hub_download(
        "PKUWilliamYang/VToonify",
        "models/shape_predictor_68_face_landmarks.dat"
    )
)


# =========================
# RUN
# =========================
morph(
    "/content/ami1.jpg",
    "/content/ma.jpg",
    landmarkpredictor,
    steps=40,
    inv_steps=300
)
````

## **Denoising Diffusion Probabilistic Models (DDPMs)**

### **DDPM forward and reverse**

````python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data

img = data.astronaut()/255.0

T = 1000

betas = np.linspace(
    1e-4,
    0.02,
    T
)

alphas = 1-betas

alpha_bar = np.cumprod(alphas)

steps = [0,50,100,250,500,750,999]

fig, ax = plt.subplots(
    1,
    len(steps),
    figsize=(18,4)
)

for i,t in enumerate(steps):

    noise = np.random.randn(*img.shape)

    xt = (
        np.sqrt(alpha_bar[t])*img
        +
        np.sqrt(1-alpha_bar[t])*noise
    )

    xt = np.clip(xt,0,1)

    ax[i].imshow(xt)
    ax[i].axis("off")
    ax[i].set_title(f"t={t}")

plt.suptitle(
    "Forward Diffusion Process"
)

plt.show()
````

````python
import torch
import numpy as np
import matplotlib.pyplot as plt
import imageio

from diffusers import DDPMPipeline

# DEVICE
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

# LOAD PRETRAINED DDPM
pipe = DDPMPipeline.from_pretrained("google/ddpm-cifar10-32")
pipe = pipe.to(device)
unet = pipe.unet
scheduler = pipe.scheduler

# START FROM PURE NOISE
sample_size = unet.config.sample_size
x = torch.randn(
    1,
    3,
    sample_size,
    sample_size,
    device=device
)
# SAVE INTERMEDIATE STATES
frames = []
capture_steps = [
    999,
    900,
    800,
    700,
    600,
    500,
    400,
    300,
    200,
    100,
    50,
    0
]
captured = {}

# REVERSE DIFFUSION
scheduler.set_timesteps(1000)

with torch.no_grad():

    for t in scheduler.timesteps:

        noise_pred = unet(
            x,
            t
        ).sample

        x = scheduler.step(
            noise_pred,
            t,
            x
        ).prev_sample

        img = (
            x[0]
            .detach()
            .cpu()
            .permute(1,2,0)
            .numpy()
        )

        img = (img + 1) / 2
        img = np.clip(img,0,1)

        frames.append(
            (img * 255).astype(np.uint8)
        )

        timestep = int(t)

        if timestep in capture_steps:
            captured[timestep] = img.copy()

# SAVE GIF
#imageio.mimsave(
#    "images/reverse_diffusion.gif",
#    frames,
#    fps=30
#)
#print("Saved reverse_diffusion.gif")

# VISUALIZE SELECTED TIMESTEPS
ordered_steps = sorted(captured.keys(), reverse=True)

fig, axes = plt.subplots(
    2,
    6,
    figsize=(15,6)
)

for ax, t in zip(
    axes.flatten(),
    ordered_steps
):

    ax.imshow(captured[t])

    ax.set_title(
        f"t={t}"
    )

    ax.axis("off")

plt.suptitle(
    "DDPM Reverse Diffusion Process",
    fontsize=16
)

plt.tight_layout()

plt.show()

fig, axes = plt.subplots(
    1,
    len(ordered_steps),
    figsize=(20,4)
)

for ax, t in zip(
    axes,
    ordered_steps
):

    ax.imshow(captured[t])

    ax.set_title(
        f"$t={t}$",
        fontsize=12
    )

    ax.axis("off")

plt.suptitle(
    "Reverse Diffusion: Noise → Image",
    fontsize=18
)

plt.show()
````

### **Latent Space Interpolation**

````python
from diffusers import AutoencoderKL
from PIL import Image
import matplotlib.pyplot as plt
import torch
from torchvision import transforms

device = "cuda"

vae = AutoencoderKL.from_pretrained(
    "stabilityai/sd-vae-ft-mse"
).to(device)

# ----------------------------------
# Load two images
# ----------------------------------

img1 = Image.open(
    "cat.png"
).convert("RGB")

img2 = Image.open(
    "dog.png"
).convert("RGB")

# ----------------------------------
# Preprocessing
# ----------------------------------

transform = transforms.Compose([

    transforms.Resize((512,512)),

    transforms.ToTensor()
])

x1 = transform(img1).unsqueeze(0).to(device)
x2 = transform(img2).unsqueeze(0).to(device)

# Stable Diffusion VAE expects [-1,1]

x1 = 2*x1 - 1
x2 = 2*x2 - 1

# ----------------------------------
# Encode
# ----------------------------------

with torch.no_grad():

    z1 = vae.encode(
        x1
    ).latent_dist.mean

    z2 = vae.encode(
        x2
    ).latent_dist.mean

print(z1.shape)


#z1 = torch.randn(1,4,64,64).cuda()
#z2 = torch.randn(1,4,64,64).cuda()

fig, axes = plt.subplots(
    1,
    8,
    figsize=(16,3)
)

for i,a in enumerate(
    torch.linspace(0,1,8)
):

    z = (1-a)*z1 + a*z2

    img = vae.decode(z).sample

    img = (img+1)/2

    img = img[0].permute(
        1,2,0
    ).cpu().detach().numpy()

    axes[i].imshow(img)
    axes[i].axis("off")

plt.show()
````

````python
from diffusers import AutoencoderKL
from PIL import Image
import matplotlib.pyplot as plt
import torch
from torchvision import transforms

device = "cuda"

vae = AutoencoderKL.from_pretrained(
    "stabilityai/sd-vae-ft-mse"
).to(device)

img = Image.open("cat.jpg").convert("RGB")

transform = transforms.Compose([
    transforms.Resize((512,512)),
    transforms.ToTensor()
])

x = transform(img).unsqueeze(0).to(device)
x = 2*x - 1

with torch.no_grad():
    latent = vae.encode(x).latent_dist.sample()

images = []

with torch.no_grad():

    for step in [999,800,600,400,200,50,0]:

        noisy = latent * (step / 1000)
        img = vae.decode(noisy).sample
        img = (img.clamp(-1,1)+1) / 2
        images.append(
            img[0]
            .permute(1,2,0)
            .cpu()
            .float()
        )

fig, axes = plt.subplots(
    1,
    len(images),
    figsize=(18,3)
)

for ax, img in zip(axes, images):

    ax.imshow(img)
    ax.axis("off")

plt.suptitle(
    "VAE Decoding of Scaled Latents"
)

plt.show()
````

````python
import torch
import matplotlib.pyplot as plt
from PIL import Image

from diffusers import (
    StableDiffusionControlNetPipeline,
    ControlNetModel,
    UniPCMultistepScheduler
)

# SETTINGS
device = "cuda"
base_model = "runwayml/stable-diffusion-v1-5"
controlnet_model = ("lllyasviel/control_v11p_sd15_openpose")

# LOAD CONTROLNET
controlnet = ControlNetModel.from_pretrained(
    controlnet_model,
    torch_dtype=torch.float16
)

# LOAD PIPELINE
pipe = StableDiffusionControlNetPipeline.from_pretrained(
    base_model,
    controlnet=controlnet,
    torch_dtype=torch.float16,
    safety_checker=None
)
pipe.scheduler = UniPCMultistepScheduler.from_config(pipe.scheduler.config)
pipe = pipe.to(device)

# MEMORY SAVING
pipe.enable_attention_slicing()
pipe.enable_vae_slicing()

# LOAD IP-ADAPTER
from diffusers.models.attention_processor import AttnProcessor2_0
pipe.unet.set_attn_processor(AttnProcessor2_0())
pipe.load_ip_adapter(
    "h94/IP-Adapter",
    subfolder="models",
    weight_name="ip-adapter_sd15.bin"
)
pipe.set_ip_adapter_scale(0.8)

# INPUTS
reference_image = Image.open("ldc2.png").convert("RGB")
openpose_image = Image.open("openpose.png").convert("RGB")
openpose_image = openpose_image.resize((512,768))

# GENERATE
images = pipe(
    prompt=
    "professional portrait photograph, "
    "highly detailed, "
    "cinematic lighting, "
    "realistic face",

    negative_prompt=
    "blurry, low quality, "
    "bad anatomy, deformed",

    image=openpose_image,
    ip_adapter_image=reference_image,
    num_inference_steps=30,
    guidance_scale=7.5,
    width=512,
    height=768,
    num_images_per_prompt=4
).images

# VISUALIZATION
fig, axes = plt.subplots(2, 3, figsize=(15,10))
axes[0,0].imshow(reference_image)
axes[0,0].set_title("Reference Identity")
axes[0,0].axis("off")
axes[0,1].imshow(openpose_image)
axes[0,1].set_title("OpenPose Control")
axes[0,1].axis("off")
axes[0,2].axis("off")
for ax, img, idx in zip(axes[1], images, range(len(images))):
    ax.imshow(img)
    ax.set_title(f"Result {idx+1}")
    ax.axis("off")
plt.suptitle("IP-Adapter + OpenPose ControlNet", fontsize=18)
plt.tight_layout()
plt.show()
````

````python
# run in Colab GPU

import os
os.environ['OPENAI_API_KEY'] = '<your API key here>'

from openai import OpenAI
import base64

client = OpenAI()

import matplotlib.pyplot as plt
from PIL import Image
import io
import requests

# ----------------- Config -----------------
OUTPUT_DIR_DALL_E_2 = "./out_dall_e_2" # only for DALL-E-2 generated images
os.makedirs(OUTPUT_DIR_DALL_E_2, exist_ok=True)

OUTPUT_DIR = "./outputs" # both DALL-E-2 and DALL-E-3 generated images side-by-side
os.makedirs(OUTPUT_DIR, exist_ok=True)
#client = OpenAI()

# ----------------- Full Prompt Bank (50) -----------------
prompts = [
    "Photo of Victoria Memorial Hall in the winter. A yellow taxi is standing in front with snows over it. On the right, a boy is taking selfie.",
    "How the pyramids of Giza would look if they were built on the surface of the moon.",
    "The streets of Venice completely dried up, with people skateboarding where gondolas once floated.",
    "Snowfall in the Sahara desert, with camels walking through white dunes.",
    "How Times Square would look during the Roman Empire, with chariots and gladiators under neon lights.",
    "The Great Wall of China glowing with neon lights at night, as if it were a futuristic city.",
    "The Taj Mahal floating in the sky among the clouds.",
    "Rio de Janeiro during a massive snowstorm.",
    "The Eiffel Tower submerged underwater with scuba divers swimming around.",
    "New York’s Central Park transformed into a giant desert with sand dunes.",
    "A tropical jungle growing inside the Colosseum in Rome.",
    "Albert Einstein as a modern-day teenage skateboarder.",
    "A female version of Leonardo da Vinci painting the Mona Lisa.",
    "Tom Cruise reimagined as a child chess prodigy.",
    "Cleopatra reimagined as a 21st-century fashion model.",
    "Shakespeare as a sci-fi film director.",
    "A school of dolphins swimming through the sky among clouds.",
    "A giraffe riding a bicycle on a busy Tokyo street.",
    "A cat leading an orchestra in a grand concert hall.",
    "How a goldfish would look if it evolved to fly like a hummingbird.",
    "An elephant surfing giant waves in Hawaii.",
    "A lion wearing a business suit giving a TED talk.",
    "A penguin hiking Mount Everest.",
    "Owls driving cars on a highway at night.",
    "A horse competing in Formula 1 racing.",
    "Kangaroos shopping in a supermarket.",
    "New York City in the year 3025, with flying cars tangled in traffic jams in the sky."]

# ----------------- Helper to download image -----------------
def get_image(url):
    response = requests.get(url)
    return Image.open(io.BytesIO(response.content))

# ----------------- DALL-E 2 generator -----------------
def generate_dalle2(prompt):
    resp_base = client.images.generate(
        model="dall-e-2",
        prompt=prompt,
        size="1024x1024",
        n=1
    )
    base_img = get_image(resp_base.data[0].url)

    resp_vars = client.images.generate(
        model="dall-e-2",
        prompt=prompt,
        size="1024x1024",
        n=2
    )
    var1 = get_image(resp_vars.data[0].url)
    var2 = get_image(resp_vars.data[1].url)

    return base_img, var1, var2

# ----------------- DALL-E 3 generator -----------------
def generate_dalle3(prompt):
    resp_base = client.images.generate(
        model="dall-e-3",
        prompt=prompt,
        size="1024x1024"
    )
    base_img = get_image(resp_base.data[0].url)

    resp_var1 = client.images.generate(
        model="dall-e-3",
        prompt=prompt,
        size="1024x1024"
    )
    var1 = get_image(resp_var1.data[0].url)

    resp_var2 = client.images.generate(
        model="dall-e-3",
        prompt=prompt,
        size="1024x1024"
    )
    var2 = get_image(resp_var2.data[0].url)

    return base_img, var1, var2

# ----------------- Run for all prompts -----------------
for idx, prompt in enumerate(prompts, start=1):
    print(f"[{idx}/{len(prompts)}] Generating for prompt: {prompt}")

    try:
        dalle2_imgs = generate_dalle2(prompt)
        dalle3_imgs = generate_dalle3(prompt)

        fig, axes = plt.subplots(1, 3, figsize=(9, 4))
        fig.suptitle(f"Prompt {idx}: {prompt}", fontsize=12, weight="bold", wrap=True)

        # DALL-E 2 row
        axes[0].imshow(dalle2_imgs[0]); axes[0].set_title("DALL·E 2 (Base)")
        axes[1].imshow(dalle2_imgs[1]); axes[1].set_title("DALL·E 2 (Var 1)")
        axes[2].imshow(dalle2_imgs[2]); axes[2].set_title("DALL·E 2 (Var 2)")

        for ax in axes:
            ax.axis("off")

        plt.tight_layout()
        out_path = os.path.join(OUTPUT_DIR_DALL_E_2, f"prompt_{idx:02d}.png")
        plt.savefig(out_path, dpi=150)
        plt.close()
        print(f"✅ Saved: {out_path}")

        fig, axes = plt.subplots(2, 3, figsize=(12, 8))
        fig.suptitle(f"Prompt {idx}: {prompt}", fontsize=12, weight="bold", wrap=True)

        # DALL-E 2 row
        axes[0, 0].imshow(dalle2_imgs[0]); axes[0, 0].set_title("DALL·E 2 (Base)")
        axes[0, 1].imshow(dalle2_imgs[1]); axes[0, 1].set_title("DALL·E 2 (Var 1)")
        axes[0, 2].imshow(dalle2_imgs[2]); axes[0, 2].set_title("DALL·E 2 (Var 2)")

        # DALL-E 3 row
        axes[1, 0].imshow(dalle3_imgs[0]); axes[1, 0].set_title("DALL·E 3 (Base)")
        axes[1, 1].imshow(dalle3_imgs[1]); axes[1, 1].set_title("DALL·E 3 (Var 1)")
        axes[1, 2].imshow(dalle3_imgs[2]); axes[1, 2].set_title("DALL·E 3 (Var 2)")

        for ax in axes.flatten():
            ax.axis("off")

        plt.tight_layout()
        out_path = os.path.join(OUTPUT_DIR, f"prompt_{idx:02d}.png")
        plt.savefig(out_path, dpi=150)
        plt.close()
        print(f"✅ Saved: {out_path}")

    except Exception as e:
        print(f"❌ Failed on prompt {idx}: {e}")
````

### **Edge-Guided Editing**

````python
import gradio as gr
import cv2
import numpy as np

def create_edge_mask(image_path):
    import cv2
    import numpy as np
    from PIL import Image

    img = cv2.imread(image_path)
    h, w = img.shape[:2]

    edges = cv2.Canny(img, 100, 200)
    edges = cv2.resize(edges, (w, h))

    mask = np.zeros((h, w, 4), dtype=np.uint8)
    mask[..., 0] = edges
    mask[..., 1] = edges
    mask[..., 2] = edges
    mask[..., 3] = edges

    Image.fromarray(mask).save("images/mask.png")
    return mask

def edge_inpaint(image):
    mask = create_edge_mask(image)    
    #mask = (edges > 0).astype(np.uint8) * 255
    
    result = client.images.edit(
        model="gpt-image-1",
        image=open(image, "rb"),
        mask=open("images/mask.png", "rb"),
        prompt="Replace edges with artistic glowing lines"
    )

    img = base64.b64decode(result.data[0].b64_json)
    return Image.open(io.BytesIO(img))

gr.Interface(
    fn=edge_inpaint,
    inputs=gr.Image(type="filepath"),
    out_dall_e_2=gr.Image(),
    title="OpenCV + GenAI Hybrid"
).launch()
````

### **Image Inpainting (Object Removal)**

````python
import numpy as np

def fix_mask(mask_path):
    mask = Image.open(mask_path).convert("L")  # grayscale
    mask_np = np.array(mask)
    h, w = mask_np.shape
    rgba = np.zeros((h, w, 4), dtype=np.uint8)
    # White → edit (transparent)
    rgba[..., 3] = 255 - mask_np   # alpha channel
    return Image.fromarray(rgba)

def inpaint(image, mask, prompt):
    if not prompt or prompt.strip() == "":
        prompt = "Fill the masked region naturally"

    # 🔥 Fix mask
    fixed_mask = fix_mask(mask)
    # fixed_mask.save("images/fixed_mask.png")

    with open(image, "rb") as img_f, open("fixed_mask.png", "rb") as mask_f:
        result = client.images.edit(
            model="gpt-image-1",
            image=img_f,
            mask=mask_f,
            prompt=prompt
        )

    import base64, io
    from PIL import Image

    img = base64.b64decode(result.data[0].b64_json)
    return Image.open(io.BytesIO(img))

gr.Interface(
    fn=inpaint,
    inputs=[
        gr.Image(type="filepath", label="Image"),
        gr.Image(type="filepath", label="Mask (white = replace)"),
        gr.Textbox(
            label="Prompt",
            value="Fill the masked region realistically"
        )
    ],
    outputs=gr.Image(),
    title="AI Inpainting Tool"
).launch()
````

### **AI Style Transfer**

````python
import gradio as gr
from openai import OpenAI
import base64
from PIL import Image
import io

# 🔑 Initialize client
client = OpenAI(api_key=api_key)

# 🔧 Prompt builder with structure strength
def build_prompt(style, strength):
    if strength < 0.3:
        return f"""
        Transform this image into {style} style.
        Feel free to creatively reinterpret the scene.
        """

    elif strength < 0.7:
        return f"""
        Apply {style} style to this image.

        Preserve general composition and main objects,
        but allow moderate artistic reinterpretation.

        Modify colors, lighting, and textures.
        """

    else:
        return f"""
        Apply {style} style to this image.

        STRICT INSTRUCTIONS:
        - Preserve composition, layout, and geometry EXACTLY
        - Do NOT change object shapes, positions, or proportions
        - Do NOT add or remove objects
        - Maintain identical perspective

        ONLY modify:
        - colors
        - textures
        - lighting

        This must look like the SAME image, only stylized.
        """


# 🔧 Main function
def style_transfer(image_path, style_choice, custom_style, strength):
    style = custom_style if custom_style.strip() else style_choice

    prompt = build_prompt(style, strength)

    result = client.images.edit(
        model="gpt-image-1",
        image=open(image_path, "rb"),
        prompt=prompt,
        size="1024x1024"
    )

    img_bytes = base64.b64decode(result.data[0].b64_json)
    return Image.open(io.BytesIO(img_bytes))


# 🎨 Style presets
styles = [
    "Studio Ghibli style (soft pastel colors, hand-painted anime, warm lighting, whimsical atmosphere)",
    "Van Gogh painting",
    "watercolor painting",
    "cyberpunk neon",
    "anime style",
    "pencil sketch",
    "oil painting",
    "photorealistic cinematic"
]


# 🖥️ Gradio UI
with gr.Blocks() as demo:
    gr.Markdown("# 🎨 AI Style Transfer with Structure Control")
    gr.Markdown("Control how much the original image is preserved")

    with gr.Row():
        with gr.Column():
            input_image = gr.Image(type="filepath", label="Upload Image")

            style_dropdown = gr.Dropdown(styles, label="Choose Style")
            custom_style = gr.Textbox(label="Or Enter Custom Style")

            structure_slider = gr.Slider(
                minimum=0.0,
                maximum=1.0,
                value=0.8,
                step=0.1,
                label="Structure Preservation (Weak → Strong)"
            )

            run_btn = gr.Button("Apply Style")

        with gr.Column():
            output_image = gr.Image(label="Stylized Output")

    run_btn.click(
        fn=style_transfer,
        inputs=[input_image, style_dropdown, custom_style, structure_slider],
        outputs=output_image
    )

demo.launch()
````

### **Detect Manufactuiring Defects**

````python
from openai import OpenAI
import base64

client = OpenAI(api_key=api_key)

IMAGE_PATH = "images/capsules.png"
with open(IMAGE_PATH, "rb") as f:
    image_b64 = base64.b64encode(f.read()).decode()

prompt = """
You are an industrial quality-control inspector.

Analyze the image carefully.

Look for:

- Scratches
- Cracks
- Corrosion
- Missing parts
- Deformation
- Surface contamination
- Paint defects
- Manufacturing anomalies

Return ONLY valid JSON:

{
  "defect_found": true,
  "defects": [
    {
      "type": "",
      "location": "",
      "severity": "",
      "confidence": 0,
      "description": ""
    }
  ],
  "overall_quality": "",
  "inspection_summary": ""
}
"""

response = client.responses.create(
    model="gpt-4o",
    input=[
        {
            "role": "user",
            "content": [
                {
                    "type": "input_text",
                    "text": prompt
                },
                {
                    "type": "input_image",
                    "image_url": f"data:image/jpeg;base64,{image_b64}"
                }
            ]
        }
    ]
)

result = response.output_text
print(result)
````

````python
import json
import cv2
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
import numpy as np

result = response.output_text.strip()
if result.startswith("```json"):
    result = result[7:]
if result.endswith("```"):
    result = result[:-3]
result = result.strip()
inspection = json.loads(result)

plt.rcParams.update({"font.size": 10, "axes.titlesize": 15, "axes.labelsize": 12})
# Load original image
img = cv2.imread(IMAGE_PATH)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
# Create dashboard
fig = plt.figure(figsize=(15, 6))
# LEFT PANEL: IMAGE
ax_img = plt.subplot(1, 2, 1)
ax_img.imshow(img)
ax_img.axis("off")
ax_img.set_title("Industrial Inspection Input", fontsize=20, fontweight="bold")
# Add numbered markers if locations
# contain ordinal descriptions
location_map = {"first": (0.12, 0.50), "second": (0.36, 0.50), "third": (0.65, 0.50), "fourth": (0.90, 0.50)} #, "rightmost": (0.88, 0.50)}

for idx, defect in enumerate(inspection["defects"], start=1):
    location = defect["location"].lower()
    for key in location_map:
        if key in location:
            x_frac, y_frac = location_map[key]
            h, w = img.shape[:2]
            x, y = int(w * x_frac), int(h * y_frac)
            ax_img.scatter(x, y, s=600)
            ax_img.text(x, y, str(idx),
                color="white", fontsize=14,
                ha="center", va="center", fontweight="bold")

# RIGHT PANEL: REPORT
ax_report = plt.subplot(1, 2, 2)
ax_report.axis("off")
quality = inspection["overall_quality"]
ax_report.text(0.02, 0.96, "QUALITY CONTROL REPORT", fontsize=22, fontweight="bold")
ax_report.text(0.02, 0.89, f"Overall Quality: {quality}", fontsize=18, fontweight="bold")
ax_report.text(0.02, 0.83, inspection["inspection_summary"], fontsize=12, wrap=True)
# Defect Cards
y = 0.70
for idx, defect in enumerate(inspection["defects"], start=1):
    severity = defect["severity"]
    if severity.lower() == "high":
        edge = "red"
    elif severity.lower() == "medium":
        edge = "orange"
    else:
        edge = "green"
    card = FancyBboxPatch((0.02, y - 0.12), 0.93,  0.13, boxstyle="round,pad=0.02", linewidth=2, edgecolor=edge, facecolor="whitesmoke")
    ax_report.add_patch(card)
    ax_report.text(0.05, y, f"#{idx} {defect['type']}", fontsize=12, fontweight="bold")
    ax_report.text(0.32, y, f"Severity: {severity}", fontsize=11)
    ax_report.text(0.65, y, f"Confidence: {100*defect['confidence']:.1f}%", fontsize=11)
    ax_report.text(0.05, y - 0.04, f"Location: {defect['location']}", fontsize=10)
    ax_report.text(0.05, y - 0.08, defect["description"], fontsize=9)
    # confidence bar
    ax_report.barh(y - 0.015, defect["confidence"], height=0.012)
    y -= 0.18
# Footer
ax_report.text(0.02, 0.05, f"Defects Detected: {len(inspection['defects'])}", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig("industrial_inspection_dashboard.png", dpi=300, bbox_inches="tight")
plt.show()
````

### **Change Detection (Remote Sensing)**

````python
# https://earthengine.google.com/timelapse - construction of bay area bridge
import base64
import json
import re
from openai import OpenAI
from PIL import Image
import matplotlib.pyplot as plt
import pandas as pd

# ----------------------------------------------------
# OpenAI Client
# ----------------------------------------------------

# ----------------------------------------------------
# Helper
# ----------------------------------------------------

def encode_image(path):
    with open(path, "rb") as f:
        return base64.b64encode(f.read()).decode()

# ----------------------------------------------------
# Input Images
# ----------------------------------------------------

before_image = "images/bay1.png"
after_image = "images/bay2.png"

before_b64 = encode_image(before_image)
after_b64 = encode_image(after_image)

# ----------------------------------------------------
# Prompt
# ----------------------------------------------------

prompt = """
You are a remote sensing analyst.

Compare Image 1 and Image 2.

Identify:

- new buildings
- removed buildings
- road changes
- new bridge
- vegetation changes
- agricultural changes
- water body changes
- urban expansion
- deforestation
- construction activity

Return ONLY valid JSON.

{
  "summary": "",
  "changes": [
    {
      "category": "",
      "description": "",
      "confidence": 0.0
    }
  ]
}
"""

# ----------------------------------------------------
# OpenAI Request
# ----------------------------------------------------

response = client.responses.create(
    model="gpt-5.1",
    input=[
        {
            "role": "user",
            "content": [

                {
                    "type": "input_text",
                    "text": prompt
                },

                {
                    "type": "input_image",
                    "image_url":
                    f"data:image/jpeg;base64,{before_b64}"
                },

                {
                    "type": "input_image",
                    "image_url":
                    f"data:image/jpeg;base64,{after_b64}"
                }

            ]
        }
    ]
)

result = response.output_text

print(result)

# ----------------------------------------------------
# Parse JSON
# ----------------------------------------------------

result = re.sub(
    r"^```json|```$",
    "",
    result.strip(),
    flags=re.MULTILINE
)

report = json.loads(result)

print("\nSummary:")
print(report["summary"])

# ----------------------------------------------------
# Convert to DataFrame
# ----------------------------------------------------

df = pd.DataFrame(
    report["changes"]
)

print("\nDetected Changes:")
df
````

### **Vision-Guided Generative Design of Drug-Like Molecules**

````python
from openai import OpenAI
import base64

# -----------------------------
# 1. Load molecule image
# -----------------------------
image_path = "images/amox.png"

def encode_image(path):
    with open(path, "rb") as f:
        return base64.b64encode(f.read()).decode("utf-8")

img_b64 = encode_image(image_path)

# -----------------------------
# 2. Use vision model to interpret molecule
# -----------------------------
data_url = f"data:image/png;base64,{img_b64}"

vision_response = client.responses.create(
    model="gpt-4.1-mini",
    input=[
        {
            "role": "user",
            "content": [
                {
                    "type": "input_text",
                    "text": (
                        "This is a chemical structure of a drug molecule. "
                        "Describe it for drug design and propose a similar novel molecule."
                    )
                },
                {
                    "type": "input_image",
                    "image_url": data_url   # ✅ FIXED HERE
                }
            ]
        }
    ]
)

print(vision_response.output_text)
analysis_text = vision_response.output_text
print("\n=== Model Interpretation ===\n")
print(analysis_text)

# -----------------------------
# 3. Convert description → new molecule image
# -----------------------------
image_prompt = f"""
A highly detailed 2D chemical structure diagram of a novel drug-like molecule.

Base inspiration:
{analysis_text}

Rules:
- Must look like a realistic medicinal chemistry structure
- Clean bond-line notation (skeletal formula)
- White background
- Black bonds with standard organic chemistry conventions
- Include heterocycles, aromatic rings, and functional groups typical of drug-like molecules
- Enforce correct valencies of the atoms participating in bonds
- No text labels
"""

image_response = client.images.generate(
    model="gpt-image-1",
    prompt=image_prompt,
    size="1024x1024"
)

image_base64 = image_response.data[0].b64_json

# -----------------------------
# 4. Save output image
# -----------------------------
output_path = "images/generated_novel_molecule.png"

with open(output_path, "wb") as f:
    f.write(base64.b64decode(image_base64))

print(f"\nSaved generated molecule image to: {output_path}")
````
